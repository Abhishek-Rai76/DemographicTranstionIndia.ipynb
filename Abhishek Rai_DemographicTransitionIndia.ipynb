## 1. EXECUTIVE SUMMARY

### Key Context
India stands at a pivotal junction in its demographic trajectory. Having officially overtaken China to become the world's most populous nation with approximately 1.44 billion citizens, India accounts for nearly 17.8% of the global population residing on just 2.4% of the world's land surface.

### Critical Finding
**India's Total Fertility Rate (TFR) has dropped to 1.9 nationally—below the replacement threshold of 2.1. Growth is now fueled entirely by demographic momentum rather than high reproduction, creating a finite 20-year window to leverage the demographic dividend.**

### Policy Urgency
The demographic dividend window peaks between 2035-2045. This analysis examines North-South regional disparities, correlations between female literacy and fertility, and strategic policy recommendations.

## 7. CORRELATION ANALYSIS

```python
# ============================================================================
# SECTION 7: CORRELATION ANALYSIS
# ============================================================================

print("\n" + "="*80)
print("CORRELATION ANALYSIS: DEMOGRAPHIC INDICATORS")
print("="*80 + "\n")

# Select numeric columns for correlation
numeric_cols = ['TFR', 'Female_Literacy', 'Urbanization', 'Median_Age', 'Growth_Rate', 'Working_Age_Share']

# Calculate correlation matrix
correlation_matrix = demographics_df[numeric_cols].corr()

print("1. CORRELATION MATRIX:")
print("-" * 80)
print(correlation_matrix.round(4))

# Specific correlations with TFR
print("\n2. CORRELATIONS WITH TFR (Total Fertility Rate):")
print("-" * 80)
tfr_correlations = correlation_matrix['TFR'].sort_values(ascending=True)
for indicator, corr_value in tfr_correlations.items():
    if indicator != 'TFR':
        print(f"  {indicator:20s}: {corr_value:8.4f}")

# Detailed analysis: Female Literacy vs TFR
print("\n3. DETAILED ANALYSIS: FEMALE LITERACY vs TFR")
print("-" * 80)

# Pearson Correlation
try:
    pearson_corr, pearson_pval = pearsonr(demographics_df['Female_Literacy'], demographics_df['TFR'])
    print(f"\nPearson Correlation:")
    print(f"  Coefficient: {pearson_corr:.6f}")
    print(f"  P-value: {pearson_pval:.8f}")
    print(f"  Significance: {'***' if pearson_pval < 0.001 else '**' if pearson_pval < 0.01 else '*' if pearson_pval < 0.05 else 'Not significant'}")
    print(f"  Interpretation: {'Statistically significant' if pearson_pval < 0.05 else 'Not statistically significant'} at 5% level")
except:
    print("Could not calculate Pearson correlation")

# Spearman Correlation
try:
    spearman_corr, spearman_pval = spearmanr(demographics_df['Female_Literacy'], demographics_df['TFR'])
    print(f"\nSpearman Correlation (Rank-based):")
    print(f"  Coefficient: {spearman_corr:.6f}")
    print(f"  P-value: {spearman_pval:.8f}")
except:
    print("Could not calculate Spearman correlation")

# Linear Regression: Female Literacy -> TFR
print(f"\n4. LINEAR REGRESSION ANALYSIS: Female Literacy → TFR")
print("-" * 80)

X = demographics_df[['Female_Literacy']].values
y = demographics_df['TFR'].values

model = LinearRegression()
model.fit(X, y)

slope = model.coef_[0]
intercept = model.intercept_
r_squared = model.score(X, y)

print(f"\nRegression Equation: TFR = {intercept:.4f} + {slope:.6f} × Female_Literacy")
print(f"\nModel Performance:")
print(f"  Slope (b): {slope:.6f}")
print(f"  Intercept (a): {intercept:.4f}")
print(f"  R² Score: {r_squared:.6f}")
print(f"  Interpretation: {r_squared*100:.2f}% of TFR variance explained by Female Literacy")

# Predictions
print(f"\n5. REGRESSION PREDICTIONS:")
print("-" * 80)
literacy_values = [50, 60, 70, 80, 90]
for lit_val in literacy_values:
    predicted_tfr = intercept + slope * lit_val
    print(f"  Female Literacy {lit_val}% → Predicted TFR: {predicted_tfr:.3f}")
```
